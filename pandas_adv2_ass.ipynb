Q2. What is the difference between the functions loc and iloc in pandas.DataFrame?
loc:
Label-based: loc is used to access a group of rows and columns by labels or a boolean array.
Inclusive: When slicing, the start and the stop values are both included.
Syntax: df.loc[row_labels, column_labels]

iloc:
Integer position-based: iloc is used to access a group of rows and columns by their integer positions (i.e., purely integer-location based indexing for selection by position).
Exclusive: When slicing, the stop value is excluded.
Syntax: df.iloc[row_indices, column_indices]

Rolling Window (rolling):

Applies a function over a rolling window of a specified size.
Common methods: .mean(), .sum(), .std(), .min(), .max(), .apply(func).

Expanding Window (expanding):

Applies a function over an expanding window starting from the beginning of the data to the current point.

Useful for giving more weight to more recent observations.

import pandas as pd
from tabulate import tabulate

def calculate_statistics():
    file_path = input("Enter the file path of the CSV file containing the student data: ")

    df = pd.read_csv(file_path)

    if 'Test Score' not in df.columns:
        print("The CSV file must contain a 'Test Score' column.")
        return

    mean_score = df['Test Score'].mean()
    median_score = df['Test Score'].median()
    mode_score = df['Test Score'].mode().tolist()

    mode_str = ', '.join(map(str, mode_score))

    statistics = [
        ["Mean", round(mean_score, 2)],
        ["Median", median_score],
        ["Mode", mode_str]
    ]

    print("\n" + tabulate(statistics, headers=["Statistic", "Value"], tablefmt="grid"))

calculate_statistics()
