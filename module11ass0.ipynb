Q1. What is MongoDB? Explain non-relational databases in short. In which scenarios it is preferred to use
MongoDB over SQL databases
MongoDB is a popular, open-source, NoSQL database that uses a document-oriented data model. It stores data in flexible, JSON-like documents, which means fields can vary from document to document and data structure can be changed over time. It is designed for scalability and high performance, and it is widely used for building modern applications that require handling large volumes of data, high throughput, and low latency.

Non-relational Databases
Non-relational databases, also known as NoSQL databases, are designed to handle a wide variety of data models, including key-value, document, columnar, and graph formats. Unlike traditional relational databases (SQL databases) which use structured query language (SQL) and predefined schemas to organize data into tables, NoSQL databases are more flexible and can handle unstructured data, which makes them ideal for handling big data and real-time web applications.

Types of NoSQL Databases:
Document-oriented databases: Store data in documents (e.g., MongoDB, CouchDB).
Key-value stores: Store data as key-value pairs (e.g., Redis, DynamoDB).
Column-family stores: Store data in columns rather than rows (e.g., Cassandra, HBase).
Graph databases: Store data in nodes and edges, suitable for data with complex relationships (e.g., Neo4j, JanusGraph).
Scenarios Where MongoDB is Preferred Over SQL Databases:
Flexible Schema Design:

MongoDB's document model is very flexible and allows for the storage of complex data structures.
Ideal for applications where the data structure may evolve over time or is not strictly defined.
Scalability:

MongoDB is designed for horizontal scalability. It can distribute data across multiple servers seamlessly.
Suitable for applications that need to scale out rather than up, such as those expecting a high volume of transactions.
Big Data and High Throughput:

MongoDB can handle large volumes of data and high write loads efficiently.
Used in applications that generate a massive amount of data, like IoT devices, social networks, and real-time analytics.
Geospatial Data:

MongoDB has powerful geospatial querying capabilities, making it suitable for applications that handle location-based data.
Ideal for use cases like geolocation services, mapping, and logistics.
Content Management and Catalogs:

MongoDB's flexibility makes it a good fit for content management systems, product catalogs, and user-generated content.
Allows for easy storage and retrieval of various types of content, such as text, images, and videos.
Prototyping and Agile Development:

MongoDB's flexible schema allows developers to iterate quickly without the need to redefine schemas or migrate databases.
Favored in startup environments and for projects following agile methodologies.
Example of MongoDB Use Case:
Consider a social media application where users can create posts, upload images, and interact with each other's content. Each post might have a different structure, some with text only, some with images, some with comments, etc. MongoDB’s flexible document model would be perfect for this application, as it allows each post to be stored as a document with a different structure.

In contrast, using a SQL database would require defining a rigid schema upfront, which might be challenging to adapt as the application evolves and new features are added.?

Q2. State and Explain the features of MongoDB.
Features of MongoDB:
Document-Oriented Storage:

MongoDB stores data in BSON (Binary JSON) format, which is similar to JSON but includes additional data types.
This allows for complex hierarchical data structures and makes it easy to represent data in a way that is close to how applications handle data.
Flexible Schema:

Unlike traditional relational databases, MongoDB does not require a predefined schema.
Documents within a collection can have different fields, which provides flexibility to evolve data models without downtime.
Scalability:

MongoDB supports horizontal scaling through sharding. Sharding distributes data across multiple servers, or shards, to handle large datasets and high throughput.
It ensures that the database can grow and manage increased loads seamlessly.
Indexing:

MongoDB supports a wide variety of indexing options to improve query performance, including single field, compound, multi-key, geospatial, text, and hashed indexes.
Indexes can be created on any field in a document, enabling fast and efficient data retrieval.
Aggregation Framework:

MongoDB provides a powerful aggregation framework to perform data processing and analysis directly within the database.
It supports operations like filtering, grouping, sorting, and transforming data using an expressive query syntax.
Replication:

MongoDB supports replication through replica sets, which consist of multiple copies of data across different servers.
Replica sets provide high availability and data redundancy, ensuring that the database can withstand server failures.
High Performance:

MongoDB is designed to deliver high performance for both read and write operations.
It supports in-memory storage, efficient indexing, and optimized data access patterns.
Geospatial Queries:

MongoDB has built-in support for geospatial data and provides a range of geospatial query operators.
It can handle location-based data and perform operations like finding nearby points, distance calculations, and geo-queries.
Full-Text Search:

MongoDB includes full-text search capabilities, allowing for efficient search operations on text fields.
It supports text indexing and various search functionalities, such as stemming, tokenization, and relevance scoring.
Transactions:

MongoDB supports multi-document ACID transactions, ensuring data consistency and reliability across multiple operations.
This feature is useful for applications that require complex transactions and need to maintain data integrity.
Rich Query Language:

MongoDB provides a powerful and flexible query language that supports a wide range of query operators and expressions.
It allows for sophisticated queries, including field selection, filtering, sorting, and projections.
Integration and Ecosystem:

MongoDB has a rich ecosystem of tools and integrations, including the MongoDB Atlas cloud service, MongoDB Compass for GUI-based data exploration, and various drivers for different programming languages.
It integrates well with other technologies, making it suitable for modern application development.
