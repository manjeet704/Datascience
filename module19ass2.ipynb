Q1. What is the Filter method in feature selection, and how does it work?
Filter methods are a type of feature selection technique that evaluates the relevance of features independently of any machine learning algorithm. They assess the relationship between features and the target variable using statistical measures.   

Calculate feature scores: Each feature is assigned a score based on its correlation or dependency with the target variable.   
Rank features: Features are ranked based on their scores.
Select features: The top-ranked features are selected as the final feature set.
Common techniques:

Correlation coefficient
Chi-square test
Information gain
Fisher's score   
Q2. How does the Wrapper method differ from the Filter method in feature selection?
Wrapper methods are another type of feature selection that directly involve the machine learning algorithm in the feature selection process. Unlike filter methods, wrapper methods evaluate different combinations of features based on their performance with the chosen model.   

Key differences:

Feature	Filter Method	Wrapper Method
Independence	Independent of ML algorithm	Dependent on ML algorithm
Computational cost	Efficient	Computationally expensive
Accuracy	Generally lower accuracy	Potentially higher accuracy

Q3. What are some common techniques used in Embedded feature selection methods?
Embedded methods combine feature selection with the training process of the machine learning model. They inherently perform feature selection as part of the model building process.   

Regularization: L1 and L2 regularization (Lasso and Ridge regression)   
Decision Trees: Feature importance in Random Forest and Gradient Boosting
Neural Networks: Pruning techniques

Q4. What are some drawbacks of using the Filter method for feature selection?
Ignoring feature interactions: Filter methods consider features independently, which might overlook important interactions between features.
Suboptimal feature subsets: They might not find the optimal subset of features that maximizes model performance.
Sensitive to noise: Filter methods can be sensitive to noise in the data, leading to suboptimal feature selection.


Q5. In which situations would you prefer using the Filter method over the Wrapper method for feature selection?
You might prefer the Filter method over the Wrapper method in the following situations:

Large datasets: Filter methods are computationally more efficient for large datasets.   
Preliminary feature reduction: Filter methods can be used as a preprocessing step to reduce the feature space before applying more computationally expensive methods.   
Understanding feature importance: Filter methods provide insights into the individual importance of features.
