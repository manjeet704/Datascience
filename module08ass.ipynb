Q4. What is a generator function in python? Why yield keyword is used? Give an example of a generator
function

What is a Generator Function in Python?
A generator function in Python is a special type of function that allows you to declare a function that behaves like an iterator. It allows you to iterate over a set of values, one at a time, without storing all of them in memory at once. This is particularly useful when working with large datasets or streams of data.

Why is the yield Keyword Used?
The yield keyword is used in generator functions to produce a value and pause the function’s execution, saving its state. When the generator function is called again, it resumes execution from where it left off, continuing until it reaches another yield statement or the end of the function. Unlike return, which terminates the function, yield allows the function to produce a series of values over time.
