When evaluating the performance of your SVM regression model, especially with a polynomial kernel, and finding that Mean Squared Error (MSE) and Root Mean Squared Error (RMSE) values are very close, you should consider the following points to choose the best metric:

Comparison of MSE and RMSE
Interpretability:

RMSE is often preferred for practical interpretation because it is in the same units as the target variable. This makes it easier to understand and communicate how far off your predictions are on average.
MSE is in squared units of the target variable, which can make it less intuitive.
Sensitivity to Outliers:

Both MSE and RMSE are sensitive to outliers because they involve squaring the error terms. However, RMSE is generally preferred when you need a metric that is more interpretable in terms of actual error magnitude.
Use Case:

RMSE is preferred when you want to present or interpret the error in a more meaningful way related to the scale of the target variable.
MSE can be useful when you want to emphasize larger errors more heavily, as it penalizes larger errors more significantly due to squaring.
Recommendation
Given that MSE and RMSE values are very close, RMSE should be your preferred metric in this case. This is because RMSE provides a more intuitive measure of error magnitude that is directly in the same units as your target variable. This makes it easier to understand how well your model is performing in practical terms.
