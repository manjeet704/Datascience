Q1. What is a database? Differentiate between SQL and NoSQL databases.
A database is a structured collection of data that is stored and accessed electronically. Databases are used to manage, organize, and retrieve large amounts of information efficiently. They are managed by database management systems (DBMS), which provide tools and functions for defining, creating, querying, updating, and administering databases.

Differentiation between SQL and NoSQL databases:
SQL Databases:
Structure:
SQL databases are relational databases. They store data in tables with rows and columns, where each row represents a record and each column represents an attribute of the data.
Schema:
SQL databases have a fixed schema. The structure of the data (schema) must be defined before data can be added. Schema changes can be complex and require significant effort.
Query Language:
SQL databases use Structured Query Language (SQL) for defining and manipulating data. SQL is a powerful language used for complex queries.
Transactions:
SQL databases support ACID (Atomicity, Consistency, Isolation, Durability) properties, which ensure reliable transactions and maintain data integrity.
Examples:
MySQL, PostgreSQL, Oracle, Microsoft SQL Server.
Use Cases:
Suitable for applications requiring complex queries and transactions, such as financial systems, ERP systems, and CRM systems.
NoSQL Databases:
Structure:
NoSQL databases are non-relational databases. They can store data in various formats such as document-based, key-value pairs, wide-column stores, or graph formats.
Schema:
NoSQL databases have a dynamic schema. They allow the insertion of data without a predefined schema, making it easier to handle unstructured or semi-structured data.
Query Language:
NoSQL databases use various query languages depending on the database type. They often provide APIs or query languages tailored to the data model they use (e.g., MongoDB uses a query language similar to JSON).
Transactions:
NoSQL databases may not fully support ACID transactions but instead support eventual consistency, which allows for higher availability and scalability.
Examples:
MongoDB (document-based), Redis (key-value store), Cassandra (wide-column store), Neo4j (graph database).
Use Cases:
Suitable for applications requiring high scalability, flexibility, and handling of large volumes of unstructured data, such as social media platforms, real-time analytics, and content management systems.

Q7. Give the order of execution of SQL clauses in an SQL query.
The order of execution of SQL clauses in an SQL query is crucial to understand how queries are processed by the database engine. The logical order of execution of SQL clauses is different from the way you write them. Here is the correct logical order of execution of SQL clauses in a query:

FROM: Determines the source table(s) from which to retrieve data.
JOIN: Combines rows from two or more tables based on a related column.
WHERE: Filters rows based on specified conditions.
GROUP BY: Groups rows that have the same values in specified columns into summary rows.
HAVING: Filters groups based on specified conditions, typically used with aggregate functions.
SELECT: Determines which columns to include in the final result set.
DISTINCT: Removes duplicate rows from the result set.
ORDER BY: Sorts the result set based on specified columns.
LIMIT / OFFSET: Restricts the number of rows returned by the query and specifies which row to start from.

SELECT DISTINCT column1, aggregate_function(column2)
FROM table1
JOIN table2 ON table1.common_column = table2.common_column
WHERE condition
GROUP BY column1
HAVING aggregate_function(column2) > value
ORDER BY column1
