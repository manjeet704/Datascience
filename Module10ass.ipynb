Q1. What is an Exception in python? Write the difference between Exceptions and Syntax errors.
An exception in Python is an error that occurs during the execution of a program. When a Python script encounters a situation that it cannot handle, it raises an exception. This interrupts the normal flow of the program, and unless the exception is handled, it will cause the program to terminate with an error message.

Exceptions are typically errors that are not detected until the program is executed, and they can be caught and handled using try-except blocks.

Difference between Exceptions and Syntax Errors
Exceptions
Definition: Exceptions are errors that occur during the execution of a program. They indicate that something unexpected has happened, such as dividing by zero, accessing a non-existent file, or calling a function with incorrect arguments.
Detection: Exceptions are detected at runtime.
Handling: Exceptions can be caught and handled using try-except blocks, allowing the program to continue running or gracefully terminate.
Examples:
ZeroDivisionError: Raised when dividing by zero.
FileNotFoundError: Raised when trying to open a file that does not exist.

Q3. Which Python statements are used to catch and handle exceptions? Explain with an example.

In Python, exceptions are caught and handled using the try, except, else, and finally statements. Here's a breakdown of each statement:

try Statement
The try block is used to wrap code that might raise an exception. If an exception occurs, the code in the try block stops executing, and the flow of control is passed to the except block.

except Statement
The except block is used to catch and handle the exception. You can specify different types of exceptions to handle them differently.

else Statement
The else block is optional and runs only if no exceptions were raised in the try block.

finally Statement
The finally block is also optional and runs regardless of whether an exception was raised or not. It is typically used to release resources or perform cleanup actions.

Q5. What are Custom Exceptions in python? Why do we need Custom Exceptions? Explain with an example.
Custom exceptions in Python are user-defined exceptions that extend the base Exception class or one of its subclasses. They allow developers to create specific error types for their applications, making error handling more precise and the code more readable and maintainable.

Why Do We Need Custom Exceptions?
Clarity: Custom exceptions make it clear what kind of error occurred, providing more context compared to generic exceptions.
Modularity: They allow different parts of a program to handle different errors in specialized ways.
Debugging: Custom exceptions can include additional information that helps in diagnosing problems.
Consistency: They promote a consistent error handling strategy throughout the codebase.
