Q1. What are the three measures of central tendency?

The three measures of central tendency are:

Mean: The mean is the average of a set of numbers. It is calculated by adding up all the values in the dataset and then dividing by the total number of values.

Median: The median is the middle value in a dataset when the values are arranged in ascending or descending order. If there is an even number of values, the median is the average of the two middle values.

Mode: The mode is the value that appears most frequently in a dataset. It is the number that occurs with the highest frequency.

Q8. What do you understand about skewness in data?
Skewness in data refers to the measure of asymmetry or lack of symmetry in the distribution of values within a dataset. It indicates whether the data is skewed to the left (negatively skewed), skewed to the right (positively skewed), or symmetrically distributed (zero skewness).

Positive Skewness (Right-skewed): In a positively skewed distribution, the tail of the distribution extends to the right, and the majority of the data points are concentrated on the left side. This means that there are relatively more low values than high values, and the mean is typically greater than the median.

Negative Skewness (Left-skewed): In a negatively skewed distribution, the tail of the distribution extends to the left, and the majority of the data points are concentrated on the right side. This means that there are relatively more high values than low values, and the mean is typically less than the median.

Zero Skewness (Symmetric): In a symmetric distribution, the data is evenly distributed on both sides of the mean, and the tails of the distribution are of equal length. This means that the mean, median, and mode are all approximately equal.

Understanding skewness in data is important because it can affect the choice of statistical analysis techniques and the interpretation of results. For example, parametric statistical tests may assume that the data is normally distributed, so skewed data may require transformation or the use of non-parametric tests. Skewness can also provide insights into the underlying characteristics of the data and help identify outliers or anomalies

Q10. Explain the difference between covariance and correlation. How are these measures used in
statistical analysis
Covariance and correlation are both measures used to quantify the relationship between two variables in statistical analysis, but they have different interpretations and properties.

Covariance:
Definition: Covariance measures the degree to which two variables change together. It indicates the direction of the linear relationship between two variables.
Calculation: Covariance is calculated by taking the average of the product of the deviations of each variable from their respective means.
Range: The range of covariance can vary from negative infinity to positive infinity.
Interpretation:
Positive covariance indicates a positive relationship between the variables (as one variable increases, the other tends to increase).
Negative covariance indicates a negative relationship between the variables (as one variable increases, the other tends to decrease).
Covariance close to zero indicates no linear relationship between the variables.
Units: Covariance is in the units of the product of the units of the two variables.
Correlation:
Definition: Correlation measures the strength and direction of the linear relationship between two variables. It standardizes the covariance by dividing it by the product of the standard deviations of the two variables.
Calculation: Correlation coefficient (Pearson correlation coefficient) is calculated as the covariance divided by the product of the standard deviations of the two variables.
Range: The range of correlation coefficient is from -1 to 1.
Interpretation:
A correlation coefficient of +1 indicates a perfect positive linear relationship.
A correlation coefficient of -1 indicates a perfect negative linear relationship.
A correlation coefficient close to 0 indicates no linear relationship between the variables.
Scale: Correlation is unitless and scale-invariant, making it easier to interpret than covariance.
Strength: Correlation measures the strength of the relationship, whereas covariance measures the direction and strength together.
Usage in Statistical Analysis:
Covariance: Covariance is used to understand the direction of the relationship between two variables. However, it is not a standardized measure, so it can be difficult to interpret the magnitude of the relationship.
Correlation: Correlation is widely used because it provides a standardized measure of the relationship between two variables. It is often used in regression analysis, hypothesis testing, and model building to assess the strength and direction of the relationship between variables.
In summary, while covariance and correlation both measure the relationship between two variables, correlation provides a more standardized and interpretable measure, making it more commonly used in statistical analysis.?

Mean: The mean is the average of all the values. For a normal distribution, this is the point around which the data is evenly distributed.

Median: The median is the middle value when the data is ordered. For a normal distribution, half of the values lie below the median and half lie above it. Since the distribution is symmetric, the median is at the center.

Mode: The mode is the value that appears most frequently. For a normal distribution, the highest point on the curve (where the frequency is the greatest) is at the center of the distribution.
