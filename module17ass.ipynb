Q4. What are the properties of Binomial distribution? Give two examples of events where binomial
distribution can be applied.
The binomial distribution is a discrete probability distribution that describes the number of successes in a fixed number of independent Bernoulli trials, each with the same probability of success. Here are the key properties and characteristics of the binomial distribution:

### Properties of Binomial Distribution

1. **Fixed Number of Trials (n):** The number of trials (or experiments) is fixed in advance.
2. **Binary Outcomes:** Each trial has only two possible outcomes: success (often denoted by \(1\)) or failure (often denoted by \(0\)).
3. **Constant Probability (p):** The probability of success is the same for each trial.
4. **Independence:** The outcome of any trial is independent of the outcome of any other trial.
5. **Discrete Distribution:** The binomial distribution is discrete, meaning it only takes on integer values.

### Probability Mass Function (PMF)

The probability of observing exactly \( k \) successes in \( n \) trials is given by the binomial formula:

\[ P(X = k) = \binom{n}{k} p^k (1-p)^{n-k} \]

where:
- \( \binom{n}{k} \) is the binomial coefficient, representing the number of ways to choose \( k \) successes out of \( n \) trials.
- \( p \) is the probability of success on any given trial.
- \( (1-p) \) is the probability of failure.

Mean (Expected Value): \( \mu = np \)
Variance:\( \sigma^2 = np(1-p) \)
Standard Deviation: \( \sigma = \sqrt{np(1-p)} \)

1. Coin Tossing:
   - Suppose you toss a fair coin 10 times. The number of heads (successes) follows a binomial distribution with \( n = 10 \) and \( p = 0.5 \).

2. Quality Control:
   - In a factory producing light bulbs, each bulb has a 2% chance of being defective. If you inspect 100 bulbs, the number of defective bulbs follows a binomial distribution with \( n = 100 \) and \( p = 0.02 \).

3. Survey Responses:
   - In a survey, 60% of respondents favor a particular policy. If you randomly select 20 respondents, the number who favor the policy follows a binomial distribution with \( n = 20 \) and \( p = 0.6 \).
4. Medical Trials:
   - In a clinical trial, a new drug has a 70% chance of curing a disease. If 15 patients are treated with the drug, the number of cured patients follows a binomial distribution with \( n = 15 \) and \( p = 0.7 \).

These examples illustrate the broad applicability of the binomial distribution in various real-world scenarios where the conditions align with its defining properties.

def poisson_cdf(mean, k):

  cdf = 0
  for i in range(k + 1):
    cdf += math.exp(-mean) * (mean ** i) / math.factorial(i)

Q7. How Binomial distribution different from Poisson distribution?
Binomial Distribution
Fixed number of trials: The experiment is repeated a fixed number of times (n).   
Two possible outcomes: Each trial results in either success or failure.   
Independent trials: The outcome of one trial does not affect the outcome of others.   
Constant probability of success: The probability of success (p) remains the same for each trial.   
Example: Flipping a coin 10 times and counting the number of heads.

Poisson Distribution
Random number of events: The number of events is not fixed, but occurs randomly over a specific interval (time, area, volume).
Rare events: The probability of an event occurring in a small interval is very small, but the total number of events can be significant.
Independent events: Events occur independently of each other.   
Example: The number of customers arriving at a store in an hour.

Q10. In normal distribution with respect to mean position, where does the least frequent data appear?
The normal distribution is symmetrical and bell-shaped, with the highest frequency of data points clustered around the mean. As you move away from the mean in either direction, the frequency of data points decreases, forming the tails of the distribution.
On a standard normal distribution curve, the data points farthest from the mean (either to the left or to the right) are the least frequent.
This means the least frequent data points are those that are several standard deviations away from the mean, in the extreme left and right ends of the bell curve.
