Q1. What is Flask Framework? What are the advantages of Flask Framework?
Flask is a lightweight and flexible web framework for Python that is designed to make it easy to build web applications quickly and with minimal overhead. It is classified as a microframework because it does not include many built-in tools and libraries, but instead provides the essential components needed to get a web application up and running. Flask is based on WSGI (Web Server Gateway Interface) and Jinja2 templating engine.

Advantages of Flask Framework
Lightweight and Minimalistic:

Flask follows a minimalistic approach, providing only the essential components to build web applications. This makes it simple and easy to understand, and developers can add only the components they need.
Flexibility:

Flask does not enforce any dependencies or project layout. This flexibility allows developers to choose the libraries and tools that best fit their needs and preferences.
Modular and Extensible:

Flask supports a modular approach, allowing developers to use various extensions to add functionality such as database integration, authentication, and form handling. Popular extensions include Flask-SQLAlchemy and Flask-Login.
Easy to Learn:

The simplicity of Flask's design makes it easy for beginners to learn and get started with web development. The straightforward API and clear documentation help developers quickly understand how to use Flask.
Built-in Development Server:

Flask includes a built-in development server that allows developers to test their applications locally without the need for an external web server.
Jinja2 Templating:

Flask uses Jinja2 as its default templating engine, which provides powerful and flexible templating capabilities. Jinja2 allows developers to create dynamic HTML content with ease.
Good Documentation:

Flask has comprehensive and well-organized documentation that makes it easier for developers to find information and learn how to use the framework effectively.
Strong Community Support:

Flask has a large and active community that contributes to its development and provides support through forums, tutorials, and third-party packages.
Scalability:

While Flask itself is minimalistic, it can be scaled and adapted for larger applications by using extensions and adhering to best practices for structuring larger codebases.
Integration with Other Technologies:

Flask can be easily integrated with other technologies and tools, such as databases (SQLAlchemy, MongoDB), authentication services, and APIs.

Q3. What is App routing in Flask? Why do we use app routes?
App routing in Flask refers to the mechanism by which you define the URL paths in your web application and associate them with specific functions or views. These routes determine how the web application responds to requests for specific URLs.

Why Use App Routes?
URL Mapping:

App routes allow you to map specific URLs to functions in your application. This means that when a user visits a particular URL, Flask knows exactly which function to execute.
Organization:

By defining routes, you can organize your web application into different pages or sections. Each route corresponds to a different view, making it easier to manage the flow of the application.
Dynamic URLs:

Flask allows you to create dynamic URLs using route parameters. This is useful when you need to capture values from the URL and use them in your view functions, such as displaying a user profile based on their username.
Cleaner Code:

Using routes helps separate the logic of handling different requests, leading to cleaner and more maintainable code.
User Experience:

Proper routing enhances the user experience by ensuring that URLs are intuitive and easy to navigate. It also allows for the creation of SEO-friendly URLs.

Q5. What function is used in Flask for URL Building? Write a Python code to demonstrate the working of the
url_for() function.
In Flask, the url_for() function is used for URL building. It generates a URL for the specified endpoint and optionally accepts arguments to build a query string.

Here is a Python code example to demonstrate the working of the url_for() function:

from flask import Flask, url_for

app = Flask(__name__)

@app.route('/')
def home():
    return 'This is the home page.'

@app.route('/welcome')
def welcome():
    return 'Welcome to ABC Corporation'

@app.route('/user/<username>')
def show_user_profile(username):
    return f'User: {username}'

@app.route('/post/<int:post_id>')
def show_post(post_id):
    return f'Post ID: {post_id}'

@app.route('/build-urls')
def build_urls():
    home_url = url_for('home')
    welcome_url = url_for('welcome')
    user_url = url_for('show_user_profile', username='john_doe')
    post_url = url_for('show_post', post_id=42)

    urls = f'''
    Home URL: {home_url}<br>
    Welcome URL: {welcome_url}<br>
    User URL: {user_url}<br>
    Post URL: {post_url}
    '''
    return urls

if __name__ == '__main__':
    app.run(debug=True)
