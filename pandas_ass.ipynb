Q1. Create a Pandas Series that contains the following data: 4, 8, 15, 16, 23, and 42. Then, print the series.

```python
import pandas as pd

data = [4, 8, 15, 16, 23, 42]
series = pd.Series(data)

print(series)
```

Q5. What are some common functions you can use to manipulate data in a Pandas DataFrame? Can you give an example of when you might use one of these functions?
Common functions:

df.head(): Displays the first few rows of the DataFrame.
df.tail(): Displays the last few rows of the DataFrame.
df.describe(): Generates descriptive statistics.
df.drop(): Removes specified labels from rows or columns.
df.fillna(): Fills NA/NaN values using the specified method.
df.groupby(): Groups DataFrame using a mapper or by a Series of columns.

data = {
    'Category': ['A', 'A', 'B', 'B'],
    'Values': [10, 20, 30, 40]
}
df = pd.DataFrame(data)
grouped_df = df.groupby('Category').mean()
print(grouped_df)
