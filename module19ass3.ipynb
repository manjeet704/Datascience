Q1. What is Min-Max scaling, and how is it used in data preprocessing? Provide an example to illustrate its application.
Min-Max scaling is a data preprocessing technique that rescales features to a specific range, typically between 0 and 1. It's useful when features have different scales and we want to bring them to a common scale.   
X_scaled = (X - X_min) / (X_max - X_min)
X is the original value
X_min is the minimum value of the feature
X_max is the maximum value of the feature
Example:
Imagine a dataset with two features: age (ranging from 18 to 80) and income (ranging from 20,000 to 200,000). Without scaling, age might have a smaller impact on the model compared to income due to their different scales. Min-Max scaling would transform both features to a range of 0 to 1, giving them equal importance.

Q2. What is the Unit Vector technique in feature scaling, and how does it differ from Min-Max scaling? Provide an example to illustrate its application.
Unit vector scaling, also known as normalization, scales features to have a length of 1. This technique is often used when the magnitude of the feature vector is important, such as in text classification or image processing.   

X_scaled = X / ||X||

X is the original feature vector
||X|| is the Euclidean norm of X
Difference from Min-Max scaling:

Min-Max scaling preserves the original distribution shape, while unit vector scaling changes the distribution.
Min-Max scaling scales features to a specific range, while unit vector scaling scales features to a unit length.
Example:
Consider a dataset of customer purchase history with features like item price, quantity, and discount. If we want to find similar customers based on their purchase behavior, unit vector scaling can be applied to each customer's purchase vector to compare them based on the direction of the vector rather than the magnitude.

Q3. What is PCA (Principal Component Analysis), and how is it used in dimensionality reduction? Provide an example to illustrate its application.
PCA is a statistical technique used to reduce the dimensionality of a dataset while preserving as much information as possible. It transforms the original features into a new set of uncorrelated features called principal components.

Calculate the covariance matrix of the data.
Compute the eigenvectors and eigenvalues of the covariance matrix.
Sort eigenvectors by decreasing eigenvalues.
Select the top 'k' eigenvectors to form a projection matrix.
Transform the original data using the projection matrix to obtain the reduced-dimensionality representation.
Example:
Imagine a dataset with 100 features representing different characteristics of a product. PCA can be used to reduce the number of features to, say, 10, while still capturing most of the variation in the data. This can be helpful for visualization or improving the performance of machine learning models.

Q4. What is the relationship between PCA and Feature Extraction, and how can PCA be used for Feature Extraction? Provide an example to illustrate this concept.
Feature extraction is the process of creating new features from existing ones. PCA is a feature extraction technique that creates new features (principal components) that are linear combinations of the original features.

Relationship between PCA and Feature Extraction:

PCA is a specific method for feature extraction.
It aims to find the directions of maximum variance in the data.
The new features (principal components) are uncorrelated and ordered by decreasing variance.
Example:
In image processing, each image can be represented as a vector of pixel values. PCA can be applied to reduce the dimensionality of the image data by extracting the most important features, leading to efficient image compression and representation.
