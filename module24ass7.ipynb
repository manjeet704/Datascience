### Q1. What Is Gradient Boosting Regression?
**Gradient Boosting Regressi** is an ensemble technique used for regression tasks that builds a model sequentially by fitting new models to the residual errors of the previous models. The core idea is to iteratively correct the mistakes made by the ensemble of models. The algorithm combines the predictions of multiple weak learners (often decision trees) to produce a strong predictive model.

### Key Concepts:

1. **Sequential Learning**: New models are trained to predict the residuals (errors) of the combined predictions of previous models.
2. **Loss Function**: The algorithm optimizes a loss function (e.g., mean squared error) by minimizing the residual errors.
3. **Learning Rate**: A parameter that controls the contribution of each new model to the ensemble.
4. **Weak Learners**: Typically shallow decision trees are used as weak learners.

### Q4. Whats a Weak Leaer in Gradient Boosting?

A **weak learner** is a model that performs slightly better than random guessing. In the context of gradient boosting, weak learners are typically simple models that are trained to capture errors from the previous models. The most common weak learners used in gradient boosting are shallow decision tree (also known as stumps) with limited depth.
