Q1. Explain why we have to use the Exception class while creating a Custom Exception.
Note: Here Exception class refers to the base class for all the exceptions
When creating a custom exception in Python, it is important to inherit from the base Exception class (or one of its subclasses) for several reasons:

Standardization: Inheriting from the Exception class ensures that your custom exception behaves like a typical Python exception. This standardization means your custom exception will integrate seamlessly with the Python exception-handling framework.

Compatibility with Exception Handling Mechanisms: Python’s built-in exception-handling mechanisms, such as try, except, finally, and raise, are designed to work with exceptions that are derived from the BaseException class (of which Exception is a direct subclass). By inheriting from Exception, your custom exception will be compatible with these mechanisms.

Access to Exception Features: Inheriting from the Exception class provides your custom exception with useful methods and attributes that are defined in the base class, such as __str__ and args. This allows you to take advantage of built-in functionality for error messages and other exception features.

Hierarchical Structure: In Python, exceptions are organized into a hierarchy. Inheriting from the Exception class allows your custom exception to fit into this hierarchy, making it easier to catch specific types of exceptions while allowing more general exceptions to be caught by broader exception handlers.

Readability and Maintenance: Using a consistent approach to exception handling makes the code more readable and maintainable. Developers who are familiar with Python’s standard exceptions will find it easier to understand and work with custom exceptions that follow the same pattern.

Q5. Explain ImportError. What is ModuleNotFoundError?
ImportError
ImportError is an exception raised in Python when an import statement fails to find the module definition or when a module cannot be loaded. This error typically occurs for one of the following reasons:

The module you are trying to import does not exist.
There is an error in the module's code that prevents it from being loaded.
The module is not in the Python path (i.e., the directories where Python looks for modules).
ModuleNotFoundError
ModuleNotFoundError is a subclass of ImportError introduced in Python 3.6. It is specifically raised when a module cannot be found. This provides a clearer distinction between a module that cannot be found and other import errors, such as errors in the module's code.
