Q1: What is the difference between a t-test and a z-test? Provide an example scenario where you would use each type of test.
t-test and z-test are both statistical tests used to compare means. The primary difference lies in the knowledge of the population standard deviation.   

t-test: Used when the population standard deviation is unknown and the sample size is relatively small (typically less than 30).
z-test: Used when the population standard deviation is known or the sample size is large (typically greater than 30).
Example:

t-test: A researcher wants to compare the average IQ scores of two groups of students (experimental and control) with a sample size of 25 in each group. Since the population standard deviation of IQ scores is unknown, a t-test would be appropriate.
z-test: A quality control manager wants to determine if the average weight of a product is different from the specified weight. If the company has historical data on the population standard deviation of the product's weight, a z-test could be used.
Q2: Differentiate between one-tailed and two-tailed tests.
One-tailed test: Used when you are interested in whether the population parameter is greater than or less than a specific value. It focuses on one direction of the distribution.
Two-tailed test: Used when you are interested in whether the population parameter is different from a specific value. It considers both tails of the distribution.
Example:

One-tailed test: A researcher wants to determine if a new drug increases blood pressure.
Two-tailed test: A researcher wants to determine if there is a difference in average test scores between two teaching methods.
Q3: Explain the concept of Type 1 and Type 2 errors in hypothesis testing. Provide an example scenario for each type of error.
Type I error is rejecting a true null hypothesis. It's often referred to as a "false positive."
Type II error is failing to reject a false null hypothesis. It's often referred to as a "false negative."

Example:

Type I error: A medical test incorrectly indicates that a patient has a disease when they actually don't (false positive).
Type II error: A medical test fails to detect a disease that the patient actually has (false negative).

Q4: Explain Bayes' theorem with an example.

Bayes' theorem is a mathematical formula used to calculate conditional probabilities. It helps update the probability of an event based on new evidence.

Formula:

P(A|B) = (P(B|A) * P(A)) / P(B)
Where:

P(A|B) is the probability of A given B
P(B|A) is the probability of B given A
P(A) is the probability of A (prior probability)   
P(B) is the probability of B
Example:
Imagine you have a medical test for a disease. The test is 95% accurate for people with the disease (true positive) and 90% accurate for people without the disease (true negative). If the prevalence of the disease is 1% in the population, what is the probability that a person actually has the disease if they test positive?

P(Disease|Positive) = (P(Positive|Disease) * P(Disease)) / P(Positive)
P(Disease|Positive) = (0.95 * 0.01) / [(0.95 * 0.01) + (0.1 * 0.99)]
Calculate the result to find the probability of having the disease given a positive test.

Q5: What is a confidence interval? How to calculate the confidence interval, explain with an example.
A confidence interval is a range of values that is likely to contain the true population parameter with a certain level of confidence.

For example, a 95% confidence interval for the population mean means that if we were to repeat the sampling process many times, 95% of the intervals would contain the true population mean.

Calculation:

The formula for a confidence interval for the population mean is:

Confidence interval = sample mean ± (critical value) * (standard error)
Where:

Sample mean is the average of the sample data   
Critical value is determined by the desired confidence level (e.g., 1.96 for a 95% confidence level)
Standard error is the standard deviation of the sample divided by the square root of the sample size
Example:
Let's say we want to find a 95% confidence interval for the average height of adult women in a city. We take a random sample of 100 women and find the mean height to be 65 inches with a standard deviation of 3 inches.

Critical value for a 95% confidence level is 1.96
Standard error = 3 / sqrt(100) = 0.3
Confidence interval = 65 ± (1.96 * 0.3) = (64.42, 65.58)

We can say with 95% confidence that the true average height of adult women in the city is between 64.42 and 65.58 inches.


Q6: Use Bayes' Theorem to calculate the probability of an event occurring given prior knowledge of the event's probability and new evidence. Provide a sample problem and solution.

Sure! Let's go through a sample problem to illustrate how Bayes' Theorem works.:

A certain disease affects 1% of the population. There is a test for the disease, but it is not perfect. The test has the following properties:
- If a person has the disease, the test is positive 99% of the time (true positive rate or sensitivity).
- If a person does not have the disease, the test is negative 95% of the time (true negative rate or specificity).

A person takes the test and the result is positive. What is the probability that this person actually has the dSolution:

Let's define the events:
- \(D\): The person has the disease.
- \( \neg D \): The person does not have the disease.
- \(T\): The test is positive.
- \( \neg T \): The test is negative.

We are given the following probabilities:
- \(P(D) = 0.01\) (prior probability of having the disease)
- \(P(\neg D) = 0.99\) (prior probability of not having the disease)
- \(P(T|D) = 0.99\) (probability of a positive test given the person has the disease)
- \(P(\neg T|\neg D) = 0.95\) (probability of a negative test given the person does not have the disease)

We need to find \(P(D|T)\), the probability that the person has the disease given a positive test result.

Using Bayes' Theorem:
\[ P(D|T) = \frac{P(T|D) \cdot P(D)}{P(T)} \]

First, we need to calculate \(P(T)\), the total probability of a positive test result:
\[ P(T) = P(T|D) \cdot P(D) + P(T|\neg D) \cdot P(\neg D) \]

Where \(P(T|\neg D) = 1 - P(\neg T|\neg D) = 1 - 0.95 = 0.05\) (probability of a positive test given the person does not have the disease).

Now calculate \(P(T)\):
\[ P(T) = (0.99 \cdot 0.01) + (0.05 \cdot 0.99) \]
\[ P(T) = 0.0099 + 0.0495 \]
\[ P(T) = 0.0594 \]

Now, we can calculate \(P(D|T)\):
\[ P(D|T) = \frac{P(T|D) \cdot P(D)}{P(T)} \]
\[ P(D|T) = \frac{0.99 \cdot 0.01}{0.0594} \]
\[ P(D|T) = \frac{0.0099}{0.0594} \]
\[ P(D|T) \approx 0.1667 \]

So, the probability that the person actually has the disease given a 
positive test result is approximately 16.67%.

Q8: What is the margin of error in a confidence interval? How does sample size affect the margin of error?
Provide an example of a scenario where a larger sample size would result in a smaller margin of error.

Margin of error is the maximum amount by which the sample statistic (like the sample mean) is expected to differ from the true population parameter. It's a measure of the precision of the estimate.   

Relationship with sample size: The margin of error is inversely proportional to the square root of the sample size. As the sample size increases, the margin of error decreases. This means that a larger sample size leads to a more precise estimate.   

Example:
In opinion polls, a larger sample size will result in a smaller margin of error, making the results more reliable. For instance, a poll with a sample size of 1000 people will have a smaller margin of error than a poll with a sample size of 500 people.

Q9: Calculate the z-score for a data point with a value of 75, a population mean of 70, and a population standard deviation of 5. Interpret the results.
Z-score = (X - μ) / σ

Where:

X = data point (75)
μ = population mean (70)
σ = population standard deviation (5)
Z-score = (75 - 70) / 5 = 1

Interpretation: The data point 75 is one standard deviation above the population mean.

Q10: In a study of the effectiveness of a new weight loss drug, a sample of 50 participants lost an average of 6 pounds with a standard deviation of 2.5 pounds. Conduct a hypothesis test to determine if the drug is significantly effective at a 95% confidence level using a t-test.
Hypothesis:

Null hypothesis (H0): The drug has no effect on weight loss (mean weight loss = 0)
Alternative hypothesis (H1): The drug has a positive effect on weight loss (mean weight loss > 0)
Test statistic:

t = (sample mean - population mean) / (sample standard deviation / sqrt(sample size))
t = (6 - 0) / (2.5 / sqrt(50)) = 10.95
Degrees of freedom: df = n - 1 = 50 - 1 = 49

Critical value: Using a t-table for a one-tailed test with 49 degrees of freedom and a 95% confidence level, the critical value is approximately 1.677.

Decision: Since the calculated t-value (10.95) is greater than the critical value (1.677), we reject the null hypothesis.

Conclusion: There is sufficient evidence to conclude that the drug is significantly effective in causing weight loss at the 95% confidence level.
