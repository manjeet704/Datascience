Q2. Differentiate between Abstraction and Encapsulation. Explain with an example.

Abstraction and Encapsulation are two fundamental concepts in Object-Oriented Programming (OOP), but they serve different purposes and operate at different levels of granularity.

Abstraction
Purpose: Abstraction is used to hide complex implementation details and show only the essential features of an object. It focuses on what an object does rather than how it does it.
Level: Abstraction works at the design level.
Implementation: It is achieved using abstract classes and interfaces.
Example: A Vehicle class that defines methods like start_engine and stop_engine without implementing the actual logic.
Encapsulation
Purpose: Encapsulation is used to bundle data (attributes) and methods (functions) that operate on the data into a single unit or class. It restricts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data.
Level: Encapsulation works at the implementation level.
Implementation: It is achieved using access specifiers like private, protected, and public.
Example: A Car class that has private attributes like engine_status and public methods to access and modify these attributes.

Q3. What is abc module in python? Why is it used?
The abc module in Python stands for "Abstract Base Classes." This module provides the infrastructure for defining abstract base classes in Python. An abstract base class (ABC) is a class that cannot be instantiated and is typically used as a blueprint for other classes.

Purpose of the abc Module
Defining Abstract Methods: The abc module allows you to define abstract methods in a class. These methods are meant to be overridden in derived classes. If a derived class does not implement these methods, it cannot be instantiated.
Creating a Framework: The abc module is used to create a common interface for a group of related classes. It ensures that any derived class implements certain methods, providing a consistent API.
Enforcing Implementation: By using abstract methods, you can ensure that certain methods are implemented in subclasses, thereby enforcing a certain design across the codebase.
How to Use the abc Module

Import the abc Module: Import the ABC class and the abstractmethod decorator from the abc module.
Define an Abstract Base Class: Create a class that inherits from ABC.
Define Abstract Methods: Use the @abstractmethod decorator to define abstract methods in the abstract base class.

Q5. Can we create an instance of an abstract class? Explain your answer.
No, we cannot create an instance of an abstract class in Python.

Explanation
Abstract classes are designed to be a blueprint for other classes rather than stand-alone objects. They can contain abstract methods, which are methods declared without any implementation. These abstract methods must be implemented by any subclass that inherits from the abstract class.

Attempting to create an instance of an abstract class will raise a TypeError because the abstract class is incomplete by itself; it needs its abstract methods to be implemented in a subclass.
