Q1: What is the difference between Ordinal Encoding and Label Encoding? Provide an example of when you might choose one over the other.
Ordinal Encoding:

Ordinal encoding is used when the categorical variables have a natural order or ranking.
It assigns a unique integer value to each category, preserving the order.
Label Encoding:

Label encoding assigns a unique integer to each category but does not imply any order or ranking.
It is used for nominal data where the categories do not have an inherent order.
Example of Usage:

Ordinal Encoding: Suppose you have a dataset with the feature "Education Level" with categories ['High School', 'Bachelor', 'Master', 'PhD']. These categories have a natural order, so ordinal encoding is appropriate.

Target Guided Ordinal Encoding:

This encoding method orders the categories based on the mean of the target variable for each category.
It is useful when you want to leverage the relationship between a categorical feature and the target variable.
How It Works:

Calculate the mean of the target variable for each category in the categorical feature.
Order the categories based on these mean values.
Assign ordinal labels to the categories based on the ordering.

Q3: Define covariance and explain why it is important in statistical analysis. How is covariance calculated?
Covariance is a statistical measure that indicates the relationship between two variables. It shows how much two variables change together.   

Importance in Statistical Analysis
Covariance is crucial in statistical analysis for several reasons:

Understanding relationships: It helps determine if two variables are positively correlated (move in the same direction), negatively correlated (move in opposite directions), or independent (no relationship).   
Portfolio management: In finance, covariance is used to assess the risk of a portfolio by measuring how asset returns move together.   
Data analysis: It's a building block for other statistical measures like correlation and variance.
Calculating Covariance
The formula for covariance between two variables X and Y is:

Cov(X, Y) = Σ[(Xi - X̄)(Yi - Ȳ)] / (n - 1)

Xi and Yi are individual data points of X and Y, respectively.
X̄ and Ȳ are the means of X and Y, respectively.
n is the number of data points.   
Steps involved:

Calculate the mean of X and Y.
Subtract the mean of X from each X value and the mean of Y from each Y value.
Multiply the corresponding differences for each data point.
Sum up all the products from step 3.
Divide the sum by (n - 1) to get the covariance.
Note:

A positive covariance indicates a positive relationship between the variables.   
A negative covariance indicates a negative relationship.   
A covariance of zero indicates no linear relationship between the variables.
